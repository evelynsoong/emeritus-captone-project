```python
import pandas as pd
import numpy as np
from sklearn.model_selection import train_test_split
from sklearn import preprocessing
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Dense, Dropout, BatchNormalization
from sklearn.metrics import mean_absolute_error,mean_squared_error,accuracy_score,r2_score
import matplotlib.pyplot as plt
from tensorflow.keras import optimizers,regularizers
import tensorflow
import joblib
import shap
```

```python
try:
    # #load dataset that has already been cleaned, feature engineered and proxy accident risk label added
    df = pd.read_csv(r"C:\Users\symev\OneDrive\Training\Emeritus\capstone project\emeritus captone project\Metro_Interstate_Traffic_Volume_Cleaned_FeatureEngineering_wProxy.csv")

    # Get number of rows and columns
    rows, columns = df.shape

    # Log successful data loading
    logger.info(
        "Raw CSV file loaded successfully: %d rows and %d columns.",
        rows,
        columns
    )

    # Display confirmation and first few rows
    print(f"\nFile loaded successfully!")
    print(f"Rows: {rows}")
    print(f"Columns: {columns}")
    print("\nFirst 5 rows:")
    print(df.head())

except FileNotFoundError:
    logger.error(f"File not found: '{file_path}'.")

except pd.errors.EmptyDataError:
    logger.error("The CSV file is empty.")

except pd.errors.ParserError:
    logger.error("The CSV file could not be parsed. Please check its format.")

except PermissionError:
    logger.error(f"Permission denied when trying to access '{file_path}'.")

except OSError as e:
    logger.error(f"Error reading the file: {e}")
```

```python
#dataset
X = df[['rain_1h', 'snow_1h', 'clouds_all', 'hour',
       'day_of_week', 'is_weekend', 'hour_sin', 'hour_cos','day_of_week_sin',
       'day_of_week_cos', 'holiday_flag',
       'weather_Clear', 'weather_Clouds', 'weather_Drizzle', 'weather_Fog',
       'weather_Haze', 'weather_Mist', 'weather_Rain', 'weather_Smoke',
       'weather_Snow', 'weather_Squall', 'weather_Thunderstorm',
       'weather_is_clear', 'weather_is_not_clear', 'temp_scaled']].values

y = df['traffic_volume'].values
```

```python
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size = 0.2, random_state = 7)
```

```python
X_train1 = X_train
X_train1
```

```python
X_test1 = X_test
```

```python
X_train1 = preprocessing.normalize(X_train1)
```

```python
model=Sequential()
model.add(Dense(10,input_dim=25,activation='relu'))
#second hidden layer
model.add(Dense(10,activation='relu'))
#output layer
model.add(Dense(1,activation='linear'))
```

```python
model.summary()
```

```python
#### compile ANN
model.compile(optimizer='Adam',loss='mse')

#fit and display the summary
model.fit(X_train1,y_train,epochs=200,verbose=1)
```

```python
X_test1 = preprocessing.normalize(X_test1)
```

```python
y_pred=model.predict(X_test1)
```

```python
# find the MSE And MAE for the evaluation
print('MSE:',mean_squared_error(y_test,y_pred))
print('MAE:',mean_absolute_error(y_test,y_pred))
```

```python
plt.figure(figsize=(5,5))
plt.scatter(y_test,y_pred)
plt.xlabel("Actual values")
plt.ylabel("Predicted values")
plt.title("ANN-Act vs Pred")
```

```python
# Save the trained model and scaler
joblib.dump(model, "deep_learning.pkl")

logger.info("MILESTONE: Deep learning ANN Model saved successfully")
```

# SHAP 

```python
# Use feature names, in the same order as X_train
feature_names = [
    'rain_1h', 'snow_1h', 'clouds_all', 'hour',
    'day_of_week', 'is_weekend', 'hour_sin', 'hour_cos',
    'day_of_week_sin', 'day_of_week_cos', 'holiday_flag',
    'weather_Clear', 'weather_Clouds', 'weather_Drizzle',
    'weather_Fog', 'weather_Haze', 'weather_Mist',
    'weather_Rain', 'weather_Smoke', 'weather_Snow',
    'weather_Squall', 'weather_Thunderstorm',
    'weather_is_clear', 'weather_is_not_clear', 'temp_scaled'
]
# Select a small representative background sample
rng = np.random.default_rng(42)
background_idx = rng.choice(
    len(X_train),
    size=min(100, len(X_train)),
    replace=False
)

background = X_train[background_idx]

# Create the explainer
explainer = shap.GradientExplainer(model, background)

# Explain a subset of test observations
X_explain = X_test[:200]
shap_values = explainer.shap_values(X_explain)

# Handle common SHAP output formats for a single-output model
if isinstance(shap_values, list):
    shap_values = shap_values[0]

shap_values = np.asarray(shap_values).squeeze(axis=-1) \
    if np.asarray(shap_values).ndim == 3 else np.asarray(shap_values)

print("SHAP values shape:", shap_values.shape)
```

```python
shap.summary_plot(
    shap_values,
    X_explain,
    feature_names=feature_names
)
```

```python
i = 0

print("Predicted traffic volume:",
      model.predict(X_test[i:i+1], verbose=0)[0, 0])

shap.plots.waterfall(
    shap.Explanation(
        values=shap_values[i],
        base_values=float(
            np.asarray(model.predict(background, verbose=0)).mean()
        ),
        data=np.asarray(X_explain[i]),
        feature_names=feature_names
    )
)
```

```python
importance = np.abs(shap_values).mean(axis=0)

shap_summary = pd.DataFrame({
    "Feature": feature_names,
    "Mean_Absolute_SHAP": importance
}).sort_values(
    "Mean_Absolute_SHAP",
    ascending=False
)

print(shap_summary.to_string(index=False))
```

```python

# Use the same feature order as during ANN training
X_route = X.copy()

# Generate ANN predictions
X_route["predicted_traffic"] = model.predict(
    X, verbose=0
).reshape(-1)

# Convert the weekend indicator into readable labels
X_route["day_type"] = np.where(
    X_route["is_weekend"] == 1,
    "Weekend",
    "Weekday"
)

# Identify the weather category from the one-hot features
weather_cols = [
    c for c in X_route.columns
    if c.startswith("weather_")
    and c not in [
        "weather_is_clear",
        "weather_is_not_clear"
    ]
]

if weather_cols:
    X_route["weather"] = (
        X_route[weather_cols]
        .idxmax(axis=1)
        .str.replace("weather_", "", regex=False)
    )
else:
    X_route["weather"] = "Unknown"

# Summarise predicted traffic by hour and day type
hourly_traffic = (
    X_route.groupby(["day_type", "hour"])
    ["predicted_traffic"]
    .agg(["mean", "count"])
    .reset_index()
    .rename(columns={
        "mean": "avg_predicted_traffic",
        "count": "observations"
    })
)

# Keep hours supported by a reasonable number
# of observations
MIN_OBSERVATIONS = 5
hourly_traffic = hourly_traffic[
    hourly_traffic["observations"] >= MIN_OBSERVATIONS
]

# Recommend the three lowest-traffic hours
def recommend_travel(day_type):
    results = (
        hourly_traffic[
            hourly_traffic["day_type"] == day_type
        ]
        .nsmallest(3, "avg_predicted_traffic")
        .sort_values("hour")
    )

    if results.empty:
        return "Insufficient data to recommend a travel window."

    hours = results["hour"].astype(int).tolist()

    windows = [
        f"{h:02d}:00–{(h + 1) % 24:02d}:00"
        for h in hours
    ]

    best_hour = int(
        results.loc[
            results["avg_predicted_traffic"].idxmin(),
            "hour"
        ]
    )

    return (
        f"For a {day_type.lower()} journey, consider travelling "
        f"during {', '.join(windows)}. "
        f"The lowest predicted traffic is around "
        f"{best_hour:02d}:00–{(best_hour + 1) % 24:02d}:00 "
        f"based on the model's estimates."
    )

print(recommend_travel("Weekday"))
print(recommend_travel("Weekend"))

# Display the ranked travel periods
print(
    hourly_traffic.sort_values("avg_predicted_traffic")
    .head(10)
)
```
