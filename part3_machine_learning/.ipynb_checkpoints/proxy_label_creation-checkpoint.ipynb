```python
import pandas as pd
```

```python
df = pd.read_csv('C:\\Users\\symev\\OneDrive\\Training\\Emeritus\\capstone project\\emeritus captone project\\Metro_Interstate_Traffic_Volume_Cleaned_FeatureEngineering.csv')
```

```python
# Congestion category (data-driven quartiles of traffic_volume)
q1, q2, q3 = df["traffic_volume"].quantile([0.25, 0.5, 0.75]).values

def bucket(v):
    if v <= q1:
        return "Low"
    elif v <= q2:
        return "Medium"
    elif v <= q3:
        return "High"
    return "Severe"

df["congestion_category"] = df["traffic_volume"].apply(bucket)
```

```python
# One-hot encode congestion category
congestion_encoded = pd.get_dummies(
    df["congestion_category"],
    prefix="congestion_category"
)

# Add the encoded columns back to the dataframe
df = pd.concat([df, congestion_encoded], axis=1)
```

```python
df.head()
```

```python
# Proxy accident-risk label
# High risk = severe/low-visibility weather conditions occurring
# together with High/Severe congestion.
# in "high_risk" column, 1 indicates high risk and 0 indicates no high risk

high_congestion = df["congestion_category"].isin(["High", "Severe"])

risky_weather = (
    df["weather_main"].isin(["Fog", "Haze", "Smoke","Thunderstorm", "Snow"]))


df["high_risk"] = (high_congestion & risky_weather).astype(int)
```
