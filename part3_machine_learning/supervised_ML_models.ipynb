```python
#dataset for classification
X = df[['temp', 'rain_1h', 'snow_1h', 'clouds_all', 'hour',
       'day_of_week', 'is_weekend', 'hour_sin', 'hour_cos','day_of_week_sin',
       'day_of_week_cos', 'holiday_flag',
       'weather_Clear', 'weather_Clouds', 'weather_Drizzle', 'weather_Fog',
       'weather_Haze', 'weather_Mist', 'weather_Rain', 'weather_Smoke',
       'weather_Snow', 'weather_Squall', 'weather_Thunderstorm',
       'weather_is_clear', 'weather_is_not_clear', 'temp_scaled']].values

y = df['high_risk'].values
```

```python
#dataset for regression
Y = df['traffic_volume'].values
```

# Classification - logistic regression

```python
logger.info("MILESTONE 1: Training Logistic Regression Classification model")
```

```python
from sklearn.model_selection import train_test_split
x_train, x_test, y_train, y_test = train_test_split(X, y, test_size=0.2 , random_state=1)
```

```python
from sklearn.preprocessing import StandardScaler
print(x_test)
scaler = StandardScaler()
scaler.fit(x_train)

x_train = scaler.transform(x_train)
x_test = scaler.transform(x_test)
print(x_test)
```

```python
from sklearn.linear_model import LogisticRegression
# clf = LogisticRegression(random_state=1)
clf = LogisticRegression(random_state=1 , penalty='l2' , solver= 'liblinear')
clf.fit(x_train, y_train)
yp = clf.predict(x_test)
```

```python
from sklearn.metrics import classification_report, confusion_matrix

print(confusion_matrix(y_test, yp))
print(classification_report(y_test, yp))
```

```python
from sklearn.metrics import roc_auc_score, roc_curve
auc_score = roc_auc_score(y_test, yp)
print(auc_score)
```

```python
# Save the trained model and scaler
joblib.dump(clf, "classification_logistic_regression_model.pkl")
joblib.dump(scaler, "standard_scaler.pkl")

logger.info("MILESTONE: Logistic Regression Classification Model and scaler saved successfully")
```


# Classification - SVM


```python
logger.info("MILESTONE 2: Training Classification- SVM model")
```

```python
#Import svm model
from sklearn import svm

#Create a svm Classifier
clf2 = svm.SVC(kernel='poly') # Linear Kernel

#Train the model using the training sets
clf2.fit(x_train, y_train)

#Predict the response for test dataset
y_pred = clf2.predict(x_test)
```

```python
from sklearn.metrics import classification_report, confusion_matrix

print(confusion_matrix(y_test, y_pred))
print(classification_report(y_test, y_pred))
```

```python
from sklearn.metrics import roc_auc_score, roc_curve
auc_score = roc_auc_score(y_test, y_pred)
print(auc_score)
```

```python
# Save the trained model and scaler
joblib.dump(clf, "classification_SVM_model.pkl")
joblib.dump(scaler, "standard_scaler.pkl")

logger.info("MILESTONE: Classification - SVM Model and scaler saved successfully")
```

# Regression - linear 

```python
logger.info("MILESTONE 3: Training Regression - linear model")
```

```python
# Split X and y into training and test set 
X_train, X_test, y_train, y_test = train_test_split(X, Y, test_size=0.20, random_state=1)
```

```python
from sklearn.linear_model import LinearRegression
regression_model = LinearRegression()
regression_model.fit(X_train, y_train)
```

```python
predicted_price_train= regression_model.predict(X_train)
predicted_price_test= regression_model.predict(X_test)
y_test[0:10]
#this shows the real price 
```

```python
predicted_price_test[0:10] #this is the estimated prices by the regression model
```

```python
from sklearn.metrics import mean_absolute_error, mean_squared_error, r2_score

mean_absolute_error(y_train, predicted_price_train) #MAE for the training set
```

```python
mean_absolute_error(y_test, predicted_price_test)  #MAE for the testing set
```

```python
r2=r2_score(y_test,predicted_price_test)
print(r2)
```

```python
# Save the trained model and scaler
joblib.dump(regression_model, "regression_linear_model.pkl")
joblib.dump(scaler, "standard_scaler.pkl")

logger.info("MILESTONE: Regression linear model and scaler saved successfully")
```

# Regression - Random forest

```python
logger.info("MILESTONE 4: Training Regression - random forest model")
```

```python
from sklearn.ensemble import RandomForestRegressor
from sklearn.metrics import mean_absolute_error, r2_score

# Train model
rfr = RandomForestRegressor(n_estimators=100, random_state=1)
rfr.fit(X_train, y_train)

# Predict
y_pred = rfr.predict(X_test)
```

```python
# Evaluate model
mae = mean_absolute_error(y_test, y_pred)
print(mae)
```

```python
r2 = r2_score(y_test, y_pred)
print(r2)
```

```python
# Save the trained model and scaler
joblib.dump(rfr, "regression_logistic_model.pkl")

logger.info("MILESTONE: Regression random forest model saved successfully")
```
